```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

activities = [
    "Problem Identification",
    "Literature Review (20+ papers)",
    "Previous Work Analysis",
    "Research Gap Identification",
    "Problem Statement & Objectives",
    "Proposed Methodology",
    "Synthetic Data Generator",
    "BiLSTM Model Development",
    "Concept Drift Detection",
    "Experience Replay",
    "Continual Learning Integration",
    "Experimental Evaluation",
    "Dashboard Development",
    "Performance Analysis",
    "Final Documentation",
    "Final Presentation & Review",
]

# 12-week project timeline
weeks = list(range(1, 13))

ranges = {
    "Problem Identification": (1, 1),
    "Literature Review (20+ papers)": (1, 3),
    "Previous Work Analysis": (2, 3),
    "Research Gap Identification": (2, 3),
    "Problem Statement & Objectives": (3, 3),
    "Proposed Methodology": (3, 4),
    "Synthetic Data Generator": (4, 5),
    "BiLSTM Model Development": (5, 6),
    "Concept Drift Detection": (6, 7),
    "Experience Replay": (7, 8),
    "Continual Learning Integration": (8, 9),
    "Experimental Evaluation": (9, 10),
    "Dashboard Development": (10, 10),
    "Performance Analysis": (10, 11),
    "Final Documentation": (11, 12),
    "Final Presentation & Review": (12, 12),
}

fig, ax = plt.subplots(figsize=(16, 10))

for i, activity in enumerate(activities):
    start, end = ranges[activity]
    ax.barh(i, end - start + 1, left=start - 0.5, height=0.58)

ax.set_yticks(range(len(activities)))
ax.set_yticklabels(activities, fontsize=10)
ax.set_xticks(weeks)
ax.set_xlim(0.5, 12.5)
ax.set_xlabel("Project Week", fontsize=11)
ax.set_title("Updated Gantt Chart – Drift-Aware Continual LSTM for Emerging Fraud Detection",
             fontsize=15, pad=16, weight="bold")
ax.grid(axis="x", linestyle="--", alpha=0.35)
ax.invert_yaxis()

# Add phase separators
for x in [3.5, 5.5, 7.5, 9.5, 11.5]:
    ax.axvline(x, linewidth=0.8, alpha=0.25)

plt.tight_layout()

path = "/mnt/data/updated_gantt_chart_fraud_detection.png"
plt.savefig(path, dpi=220, bbox_inches="tight")
plt.close()

print(path)
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
activities = [
    "Problem Identification",
    "Literature Review (20+ papers)",
    "Previous Work Analysis",
    "Research Gap Identification",
    "Problem Statement & Objectives",
    "Proposed Methodology",
    "Synthetic Data Generator",
    "BiLSTM Model Development",
    "Concept Drift Detection",
    "Experience Replay",
    "Continual Learning Integration",
    "Experimental Evaluation",
    "Dashboard Development",
    "Performance Analysis",
    "Final Documentation",
    "Final Presentation & Review",
]
weeks = list(range(1, 9))
ranges = {
    "Problem Identification": (1, 1),
    "Literature Review (20+ papers)": (1, 2),
    "Previous Work Analysis": (1, 2),
    "Research Gap Identification": (2, 2),
    "Problem Statement & Objectives": (2, 2),
    "Proposed Methodology": (2, 3),
    "Synthetic Data Generator": (3, 4),
    "BiLSTM Model Development": (3, 4),
    "Concept Drift Detection": (4, 5),
    "Experience Replay": (4, 5),
    "Continual Learning Integration": (5, 6),
    "Experimental Evaluation": (6, 7),
    "Dashboard Development": (6, 7),
    "Performance Analysis": (7, 8),
    "Final Documentation": (7, 8),
    "Final Presentation & Review": (8, 8),
}
fig, ax = plt.subplots(figsize=(14, 8), dpi=300)
colors = [
    "#3498db", "#3498db", "#3498db", "#3498db", "#3498db", "#2980b9",
    "#2ecc71", "#27ae60",
    "#f39c12", "#e67e22", "#d35400",
    "#9b59b6", "#8e44ad",
    "#34495e", "#2c3e50", "#e74c3c"
]
for i, activity in enumerate(activities):
    start, end = ranges[activity]
    duration = end - start + 1

    ax.barh(
        i,
        duration,
        left=start - 0.45,
        height=0.55,
        color=colors[i],
        edgecolor="black",
        linewidth=0.5,
        alpha=0.9
    )
ax.set_yticks(range(len(activities)))
ax.set_yticklabels(
    activities,
    fontsize=10,
    fontweight="medium"
)
ax.set_xticks(weeks)
ax.set_xticklabels(
    [f"Week {w}" for w in weeks],
    fontsize=10,
    fontweight="bold"
)
ax.set_xlim(0.5, 8.5)
ax.set_xlabel(
    "Project Timeline (8 Weeks)",
    fontsize=11,
    fontweight="bold",
    labelpad=10
)
ax.set_title(
    "8-Week Gantt Chart – Drift-Aware Continual LSTM for Emerging Fraud Detection",
    fontsize=13,
    pad=16,
    weight="bold",
    color="#2c3e50"
)
ax.grid(
    axis="x",
    linestyle="--",
    alpha=0.4,
    color="#7f8c8d"
)
ax.invert_yaxis()

for x in [2.55, 4.55, 6.55]:
    ax.axvline(
        x,
        linewidth=1.2,
        color="#e74c3c",
        linestyle=":",
        alpha=0.6
    )
legend_elements = [
    Patch(
        facecolor="#3498db",
        edgecolor="black",
        label="Phase 1: Planning & Literature Review"
    ),
    Patch(
        facecolor="#2ecc71",
        edgecolor="black",
        label="Phase 2: Core Deep Learning Modeling"
    ),
    Patch(
        facecolor="#f39c12",
        edgecolor="black",
        label="Phase 3: Drift Detection & Continual Learning"
    ),
    Patch(
        facecolor="#9b59b6",
        edgecolor="black",
        label="Phase 4: Evaluation & Dashboard"
    ),
    Patch(
        facecolor="#34495e",
        edgecolor="black",
        label="Phase 5: Documentation & Final Review"
    )
]
ax.legend(
    handles=legend_elements,
    loc="upper center",
    bbox_to_anchor=(0.5, -0.12),
    ncol=2,
    fontsize=9,
    frameon=True,
    title="Project Phases",
    title_fontsize=10
)
plt.tight_layout()
output_path = "gantt_chart_8_weeks.png"
plt.savefig(
    output_path,
    dpi=300,
    bbox_inches="tight"
)
plt.close()
print(f"Gantt chart saved successfully as {output_path}")
```
